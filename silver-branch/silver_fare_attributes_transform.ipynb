```python
dbutils.fs.ls("abfss://bronze@grtadlsdev.dfs.core.windows.net")

df = spark.table("bronze.fare_attributes")
display(df.limit(20))
```

```python
df.printSchema()
```

```python
df = df.withColumnRenamed("transfer_duration", "transfer_duration_seconds")
```

```python
%sql
create schema if not exists silver;
```

```python
df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("silver.fare_attributes")

silver_path = "abfss://silver@grtadlsdev.dfs.core.windows.net/fare_attributes/"
df.write.format("delta").mode("overwrite").option("overwrite schema", "true").save(silver_path)
```
