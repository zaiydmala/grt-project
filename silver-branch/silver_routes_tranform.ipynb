```python
from pyspark.sql import functions as F
from pyspark.sql.types import *

dbutils.fs.ls("abfss://bronze@grtadlsdev.dfs.core.windows.net/")

# show first 100 rows
df = spark.table("bronze.routes")
display(df.limit(100))

# null counts
df.select([F.sum(F.col(c).isNull().cast("int")).alias(c) for c in df.columns]).show()
```

```python
# drop route desc
df = df.drop("route_desc")
```

```python
df.printSchema()
```

```python
silver_path = "abfss://silver@grtadlsdev.dfs.core.windows.net/routes/"

# write as delta
df.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .save(silver_path)

display(spark.read.format("delta").load(silver_path).limit(5))
```
