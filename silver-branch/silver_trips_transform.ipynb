```python
from pyspark.sql import functions as F
from pyspark.sql.types import *

dbutils.fs.ls("abfss://bronze@grtadlsdev.dfs.core.windows.net/")

df = spark.table("bronze.trips")
display(df.limit(50))

# null counts
df.select([F.sum(F.col(c).isNull().cast("int")).alias(c) for c in df.columns]).show()



```

```python
df.printSchema()
```

```python
df.filter(F.col("bikes_allowed") == 0).count()
```

```python
# make wheelchair_accessible column a boolean
df = df.withColumn("wheelchair_accessible", F.col("wheelchair_accessible").cast("boolean"))
df = df.withColumn("bikes_allowed", F.col("bikes_allowed").cast("boolean"))

df.printSchema()
```

```python
df.limit(100).display()
```

```python
%sql
create schema if not exists silver
```

```python
df.write.format("delta").mode("overwrite").option("overwriteSchema" ,"true").saveAsTable("silver.trips")

silver_path = "abfss://silver@grtadlsdev.dfs.core.windows.net/trips/"
df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").save(silver_path)
```
